```python

# Import the required libraries
import requests
from bs4 import BeautifulSoup

# Website used for our web scraping practical
url = "https://books.toscrape.com/"

# Send a request to download the webpage
response = requests.get(url, timeout=20)

# Check whether the request was successful
print("Status Code:", response.status_code)

# Display the first 500 characters of the webpage
print(response.text[:500])
```

```python

# Parse the downloaded HTML
soup = BeautifulSoup(response.text, "html.parser")

# Find all book containers on the page
books = soup.find_all("article", class_="product_pod")

# Create an empty list to store book details
book_data = []

# Extract the title and price of each book
for book in books:
    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text

    # Store the details in a dictionary
    book_data.append({
        "Title": title,
        "Price": price
    })

# Display how many books were extracted
print("Total books extracted:", len(book_data))

# Display the first five books
for item in book_data[:5]:
    print(item)
```

```python

# Import pandas for tabular data handling
import pandas as pd

# Convert the extracted book data into a DataFrame
df = pd.DataFrame(book_data)

# Display the first five rows
display(df.head())

# Display the dataset shape
print("Rows and columns:", df.shape)

# Check column names and data types
df.info()
```

```python
# Clean the Price column by removing currency symbols and extra characters
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace(r"[^\d.]", "", regex=True)
    .astype(float)
)

# Display the cleaned data
display(df.head())

# Check the updated data types
df.info()
```

```python
# Find the most expensive book
print("Most expensive book:")
display(df.loc[df["Price"].idxmax()])

# Find the cheapest book
print("\nCheapest book:")
display(df.loc[df["Price"].idxmin()])
```

```python
# Save the scraped book data to a CSV file
df.to_csv("scraped_books.csv", index=False)

print("Data saved successfully as scraped_books.csv")
```
