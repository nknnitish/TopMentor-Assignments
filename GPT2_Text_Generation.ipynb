```python
# Import the required libraries
from transformers import pipeline
import torch

# Check whether a GPU is available
if torch.cuda.is_available():
    print("GPU available")
else:
    print("CPU mode")

print("Libraries imported successfully")
```

```python
# Load the pretrained GPT-2 text-generation model
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

print("GPT-2 model loaded successfully")
```

```python
# Define the starting prompt
prompt = "Artificial Intelligence is changing the world because"

# Generate text using GPT-2
results = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8,
    pad_token_id=generator.tokenizer.eos_token_id
)

# Display the generated text
print("Generated Text:")
print(results[0]["generated_text"])
```

```python
# Generate text about a topic of your choice
prompt = "Data science is an important career because"

results = generator(
    prompt,
    max_new_tokens=80,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8,
    pad_token_id=generator.tokenizer.eos_token_id
)

# Display the result
print(results[0]["generated_text"])
```

```python
# Generate three different text continuations
prompt = "The future of artificial intelligence"

results = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=3,
    do_sample=True,
    temperature=0.9,
    pad_token_id=generator.tokenizer.eos_token_id
)

# Display all generated variations
for i, result in enumerate(results, start=1):
    print(f"\nVariation {i}:")
    print(result["generated_text"])
```

```python
# Save generated text to a text file
with open("gpt2_generated_text.txt", "w", encoding="utf-8") as file:
    for i, result in enumerate(results, start=1):
        file.write(f"Variation {i}:\n")
        file.write(result["generated_text"] + "\n\n")

print("Generated text saved successfully!")
```
